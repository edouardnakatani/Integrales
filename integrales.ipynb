```python
import matplotlib.pyplot as plt
import numpy as np

from Methodes import rectangles_gauche, rectangles_droite, trapezes, point_milieu, simpson
```

```python
def f(x):
    return 3 * x**2
```

```python
print (f'Intégrale par la méthode des rectangles à gauche : {rectangles_gauche(f, 0, 1, 2)}')
print (f'Intégrale par la méthode des rectangles à droite : {rectangles_droite(f, 0, 1, 2)}')
```

```python
n = 50
X = np.linspace(1, n, n)
Y_g = np.zeros(n)
Y_d = np.zeros(n)
for i in range(n):
    Y_g[i] = rectangles_gauche(f, 0, 1, i + 1)
    Y_d[i] = rectangles_droite(f, 0, 1, i + 1)

plt.plot(X, Y_g)
plt.plot(X, Y_d)
plt.show()
```

```python
def trapezes(f, a, b, n):
    h = (b - a) / n
    x = a
    integrale = 0
    for i in range(n):
        integrale = integrale + (f(x) + f(x + h)) / 2 * h
        x = x + h
    return integrale
```

```python
print (f'Intégrale par la méthode des trapèzes : {trapezes(f, 0, 1, 1000)}')
```
